```python
# this is the same as the append function
def stack_push(s, value):
    return s + [value]

#this is the same as the pop function
def stack_pop(s): 
    tos = s[-1]
    del s[-1]
    return tos

url_stack = []
```

```python
wikipedia_datascience = "Data science is an interdisciplinary field that uses scientific methods, processes, algorithms and system s to extract knowledge [https://en.wikipedia.org/wiki/Knowledge] and insights from data [https://en.wikipedia.org/wiki/Data] in v arious forms, both structured and unstructured, similar to data m ining [https://en.wikipedia.org/wiki/Data_mining]"
```

```python
len(wikipedia_datascience)
```

```python
wd_list = wikipedia_datascience.split() 
len(wd_list)
```

```python
for word in wd_list:
    if word.startswith("[https://"):
        url_stack = stack_push(url_stack, word[1:-1])
        
# Notice the clever use of string slicing
url_stack
```

```python
for i in range(0, len(url_stack)): 
    print(stack_pop(url_stack))
```

Lambda function

```python
import math

def my_sine():
    return lambda x: math.sin(math.radians(x))

def my_cosine():
    return lambda x: math.cos(math.radians(x))
```

```python
sine = my_sine() 
cosine = my_cosine()
math.pow(sine(60), 2) + math.pow(cosine(60), 2)
```
