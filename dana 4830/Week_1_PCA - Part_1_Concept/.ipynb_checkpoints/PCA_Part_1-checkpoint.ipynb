```python
import numpy as np
```

```python
data = np.array([
    [4, 5, 1, 5],
    [2, 3, 1, 4],
    [3, 4, 3, 2],
    [5, 4, 4, 1],
    [5, 5, 2, 3]
])
data = data[:, :2]
data
```

## Easy Version

`scikit-learn` has `PCA` class for performing Principal Component Analysis.

Reference: https://scikit-learn.org/stable/modules/generated/sklearn.decomposition.PCA.html

```python
from sklearn.decomposition import PCA
```

```python
# fit data to PCA
pca = PCA()
pca.fit(data)
```

```python
# get loadings
pca.components_
```

```python
pca.explained_variance_
```

```python
pca.explained_variance_ratio_
```

```python
# get scores
pca.transform(data)
```

## Hard Version

We can implement our PCA using numpy methods.

Steps:
1. Center the mean of data to origin
2. Calculate the covariance matrix
3. Perform eigendecomposition to obtain eigenvalues and eigenvectors
4. Sort the eigenvectors by variance (eigenvalues) to determine the ordering of Principal Components (PCs).

```python
class SimplePCA:
    '''
    Naive implementation of scikit-learn style PCA class. For demo purpose only.
    Sign ambiguity in eigendecomposition can cause non-determistic outputs.
    '''
    def __init__(self, n_components=None):
        self.explained_variance_ = None
        self.explained_variance_ratio_ = None
        self.components_ = None
        self.__n_components = n_components
    
    def __repr__(self):
        return f"{__class__.__name__}()"
    
    def fit(self, data):
        if not self.__n_components or self.__n_components > data.shape[1]:
            self.__n_components = data.shape[1]
            
        # mean_data = self._standardize(data)
        mean_data = self._mean_centering(data)
        
        # compute covariance
        cov = np.cov(mean_data.T)
        
        # perform eigendecomposition of covariance matrix
        eig_vals, eig_vecs = np.linalg.eig(cov)
        
        # sort eigenvalues and corresponding eigenvectors in descending order, get the top-n components
        indices = np.arange(0,len(eig_vals), 1)
        indices = ([x for _,x in sorted(zip(eig_vals, indices), reverse=True)])[:self.__n_components]
        self.explained_variance_ = eig_vals[indices]
        self.explained_variance_ratio_ = (eig_vals / np.sum(eig_vals))[indices]
        self.components_ = eig_vecs[:,indices].T
        return self
    
    def transform(self, data):
        mean_data = self._mean_centering(data)
        return np.dot(mean_data, self.components_.T)
    
    def fit_transform(self, data):
        self.fit(data)
        return self.transform(data)

    def _mean_centering(self, data):
        '''
        subtract data with column-wise mean (mean-centering)
        '''
        return data - np.mean(data, axis=0)
```

```python
# fit data to PCA
pca = SimplePCA()
pca.fit(data)
```

```python
# get loadings
pca.components_
```

```python
pca.explained_variance_
```

```python
pca.explained_variance_ratio_
```

```python
# get scores
pca.transform(data)
```
