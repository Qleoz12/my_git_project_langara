```python
import pandas as pd
url = 'https://raw.githubusercontent.com/pandas-dev/pandas/master/pandas/tests/io/data/csv/iris.csv'
iris_df = pd.read_csv(filepath_or_buffer=url, header=0)
iris_df.head(2)
```

```python
import sqlite3
# Create your connection.
db_conn = sqlite3.connect(':memory:')
iris_df.to_sql(name='iris', con=db_conn)
```

```python
df_from_db = pd.read_sql('select * from iris where PetalWidth > 0.2 AND Name != "Iris-setosa" ', db_conn)
df_from_db.head(2)
```
