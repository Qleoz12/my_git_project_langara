Plot Multiple Lines with Dual Axis
We put them into 2 subplots, and set the xaxis being the same.
ax1 stands for the axis for subplot graph 1
ax2 stands for the axis for subplot graph 2

Legend can be set at different location by "loc=".
You can choose: 'best','upper right', 'upper left', 'lower left', 'lower right', 'center', 'center left', 'center right', 'lower center', 'upper center'.

```python
# Sample data
x = np.arange(0, 10, 0.1)
y1 = np.exp(x)
y2 = np.log(x + 1)

# Create a figure and a set of subplots
fig, ax1 = plt.subplots()

# Plot the first line with the primary y-axis
color = 'tab:red'
ax1.set_xlabel('x (s)')
ax1.set_ylabel('exp', color=color)
line1, = ax1.plot(x, y1, color=color, label='Exp')  # Add label for the legend
ax1.tick_params(axis='y', labelcolor=color)

# Create a twin Axes sharing the x-axis
ax2 = ax1.twinx()

# Plot the second line with the secondary y-axis
color = 'tab:blue'
ax2.set_ylabel('log', color=color)
line2, = ax2.plot(x, y2, color=color, label='Log')  # Add label for the legend
ax2.tick_params(axis='y', labelcolor=color)

# Add a title
plt.title('Two Lines with Primary and Secondary Y-Axis')

# Handling legend
# Combine the legend elements from both axes
lines = [line1, line2]
labels = [line.get_label() for line in lines]
ax1.legend(lines, labels, loc='upper right')  # Change legend position here

fig.tight_layout()  # Adjust the layout
plt.show()
```

Subplot: to plot a lot of graphs side to side for easy comparison.

```python
# Create a figure and a 3x2 grid of subplots
fig, axs = plt.subplots(3, 2, figsize=(10, 15))  # figsize is optional

# Sample data for plotting
x = np.linspace(0, 2 * np.pi, 400)
y1 = np.sin(x ** 2)
y2 = np.cos(x ** 2)

# Plotting on each subplot
axs[0, 0].plot(x, y1)
axs[0, 0].set_title('Axis [0, 0]')

axs[0, 1].plot(x, y2, 'tab:orange')
axs[0, 1].set_title('Axis [0, 1]')

axs[1, 0].plot(x, -y1, 'tab:green')
axs[1, 0].set_title('Axis [1, 0]')

axs[1, 1].plot(x, -y2, 'tab:red')
axs[1, 1].set_title('Axis [1, 1]')

axs[2, 0].plot(x, y1, 'tab:purple')
axs[2, 0].set_title('Axis [2, 0]')

axs[2, 1].plot(x, y2, 'tab:brown')
axs[2, 1].set_title('Axis [2, 1]')

# Adjust layout to prevent overlap
plt.tight_layout()

# Show plot
plt.show()
```

Pie Chart

```python
# Data to plot
labels = ['Python', 'Java', 'C++', 'JavaScript']
sizes = [215, 130, 245, 210]  # The sizes of each slice
colors = ['gold', 'yellowgreen', 'lightcoral', 'lightskyblue']  # Colors for each slice
explode = (0, 0, 0., 0)  # "Explode" the 1st slice (Python)

# Plot
plt.figure(figsize=(8, 8))  # Set the figure size
plt.pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.2f%%', shadow=True, startangle=140)
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.

plt.title('Programming Language Usage')  # Title for the pie chart
plt.show()
```

```python
# Data for plotting
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

# Create a bar chart
plt.figure(figsize=(8, 8))  # Optional: Specifies the figure size
plt.bar(products, sales, color='skyblue')

# Adding titles and labels
plt.title('Sales Figures for Products')
plt.xlabel('Products')
plt.ylabel('Sales')

# Optional: Adding value labels on top of each bar
for i, v in enumerate(sales):
    plt.text(i, v + 1, str(v), ha='center', va='bottom')
    #print(str(i), str(v))

# Show the plot
plt.show()
```

```python
# Applying a predefined style
plt.style.use('ggplot')

# Data for plotting
labels = ['G1', 'G2', 'G3', 'G4', 'G5']
men_means = [20, 35, 30, 35, 27]
women_means = [25, 32, 34, 20, 25]

x = np.arange(len(labels))  # Label locations
width = 0.35  # Width of the bars

# Create a figure and an axes
fig, ax = plt.subplots()

# Plot data
rects1 = ax.bar(x - width/2, men_means, width, label='Men', color='blue')
rects2 = ax.bar(x + width/2, women_means, width, label='Women', color='orange')

# Add labels, title, and custom x-axis tick labels
ax.set_xlabel('Group')
ax.set_ylabel('Scores')
ax.set_title('Scores by group and gender')
ax.set_xticks(x)
ax.set_xticklabels(labels)

# Add a legend
ax.legend()

# Show the plot
plt.show()
```

Stacked Bar chart

```python
# Sample data
categories = ['Product A', 'Product B', 'Product C']
sales_Q1 = [10, 20, 30]  # Sales in the first quarter
sales_Q2 = [15, 25, 35]  # Sales in the second quarter

# Position of bars on the x-axis
ind = np.arange(len(categories))

# Size of the figure
plt.figure(figsize=(10, 7))

# Plotting
plt.bar(ind, sales_Q1, width=0.4, label='Q1 Sales', color='skyblue')
plt.bar(ind, sales_Q2, width=0.4, bottom=sales_Q1, label='Q2 Sales', color='orange')

# Adding titles and labels
plt.xlabel('Product Categories')
plt.ylabel('Sales')
plt.title('Sales by Product Category and Quarter')

# Adding category names to the x-axis
plt.xticks(ind, categories)

# Adding legend
plt.legend()

# Show the plot
plt.show()
```

Box and Whisker plot

```python
# Generating random data: create three sets of data
data1 = np.random.normal(100, 10, 200)  # Mean=100, SD=10, N=200
data2 = np.random.normal(80, 30, 200)   # Mean=80, SD=30, N=200
data3 = np.random.normal(90, 20, 200)   # Mean=90, SD=20, N=200

# Combine these data sets into a list
data = [data1, data2, data3]

# Creating the box plot
plt.figure(figsize=(8, 6))  # Optional: Setting the figure size
plt.boxplot(data, patch_artist=True, notch=True, labels=['Data 1', 'Data 2', 'Data 3'])
#plt.boxplot(data, patch_artist=True, notch=False, labels=['Data 1', 'Data 2', 'Data 3'])

# Adding titles and labels
plt.title('Simple Box and Whisker Plot')
plt.xlabel('Data Set')
plt.ylabel('Value')

# Show the plot
plt.show()
```

3D Plot

```python
from mpl_toolkits.mplot3d import Axes3D

# Generating random sample data
x = np.random.standard_normal(100)
y = np.random.standard_normal(100)
z = np.random.standard_normal(100)

# Creating a figure and a 3D subplot
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Scatter plot
ax.scatter(x, y, z, c='blue', marker='o', alpha=0.5)

# Adding titles and labels
ax.set_title('3D Scatter Plot Example')
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')

# Show plot
plt.show()
```

Polar chart

```python
# Sample data
r = np.arange(0, 2, 0.01)  # Radius values
theta = 2 * np.pi * r  # Angle values

# Creating a polar plot
plt.figure(figsize=(8, 6))  # Optional: Setting the figure size
ax = plt.subplot(111, projection='polar')  # Creating a subplot with polar projection
ax.plot(theta, r, color='blue')  # Plotting data on the polar subplot

# Customizing the plot
ax.set_title('Simple Polar Chart')  # Adding a title
ax.set_rmax(2)  # Setting the maximum radial limit
ax.set_rticks([0.5, 1, 1.5, 2])  # Setting the radial ticks
ax.set_rlabel_position(-22.5)  # Setting the position of the radial labels
ax.grid(True)  # Enabling the grid

# Show plot
plt.show()
```

Contour plot

```python
# Generating the data
x = np.linspace(-3.0, 3.0, 100)  # 100 linearly spaced numbers between -3 and 3
y = np.linspace(-3.0, 3.0, 100)  # 100 linearly spaced numbers between -3 and 3
X, Y = np.meshgrid(x, y)  # Create a mesh grid
Z = np.sin(np.sqrt(X**2 + Y**2))  # Function to plot

# Creating the contour plot
plt.figure(figsize=(8, 6))  # Optional: Setting the figure size
contours = plt.contour(X, Y, Z, 3, colors='black')  # Draw contour lines
plt.clabel(contours, inline=True, fontsize=8)  # Label the contours

# Filling the contours
#plt.contourf(X, Y, Z, 3, cmap='viridis')  # Add a color fill between the lines
plt.contourf(X, Y, Z, 3, cmap='hot')
plt.colorbar()  # Add a color bar to the side

# Adding titles and labels
plt.title('Simple Contour Plot')
plt.xlabel('x axis')
plt.ylabel('y axis')

# Show plot
plt.show()
```
